### ✅ How can computing descriptive statistics help you learn more about your data in this stage of your analysis?

Computing descriptive statistics is an essential part of Exploratory Data Analysis (EDA) because it provides a summary view of the dataset, helping to uncover patterns, anomalies, and initial insights. In this project, descriptive statistics can:

- **Summarize central tendency**: Metrics like mean and median help us understand the typical fare amount in the dataset.
- **Measure spread**: Standard deviation and range show how much variation there is in fare amounts, which is important when comparing groups.
- **Identify outliers**: Extremely high or low values in fare amounts or trip distances can affect the results of statistical tests, so spotting them early is crucial.
- **Compare groups**: By comparing the descriptive statistics for credit card vs. cash payments, we can see initial differences that may suggest a relationship between payment type and fare amount.
- **Guide hypothesis testing**: These summaries inform how we structure our null and alternative hypotheses, and what kind of statistical tests are most appropriate.

Overall, descriptive statistics lay the groundwork for more rigorous hypothesis testing and help ensure the integrity and focus of the analysis.

**Note:** In the dataset, `payment_type` is encoded in integers:
*   1: Credit card

```python
# Ver los primeros registros del dataset
print(taxi_data.head())

# Revisar el resumen estadístico general
print(taxi_data.describe())

# Ver la distribución de tipos de pago
print(taxi_data["payment_type"].value_counts())

# Filtrar solo los pagos con tarjeta de crédito y en efectivo
taxi_filtered = taxi_data[taxi_data["payment_type"].isin([1, 2])]

# Estadísticas descriptivas para cada grupo
credit_card_fares = taxi_filtered[taxi_filtered["payment_type"] == 1]["fare_amount"]
cash_fares = taxi_filtered[taxi_filtered["payment_type"] == 2]["fare_amount"]

# Mostrar estadísticas descriptivas para cada tipo de pago
print("📊 Credit Card Fare Stats:\n", credit_card_fares.describe())
print("\n📊 Cash Fare Stats:\n", cash_fares.describe())
```

You are interested in the relationship between payment type and the fare amount the customer pays. One approach is to look at the average fare amount for each payment type. 

```python

# Calcular el promedio de fare_amount por tipo de pago
average_fare_by_payment_type = taxi_data.groupby("payment_type")["fare_amount"].mean()

# Mostrar resultados
print("💳 Average Fare Amount by Payment Type:")
print(average_fare_by_payment_type)
```

Based on the averages shown, it appears that customers who pay in credit card tend to pay a larger fare amount than customers who pay in cash. However, this difference might arise from random sampling, rather than being a true difference in fare amount. To assess whether the difference is statistically significant, you conduct a hypothesis test.


### Task 3. Hypothesis testing

Before you conduct your hypothesis test, consider the following questions where applicable to complete your code response:

1. Recall the difference between the null hypothesis and the alternative hypotheses. Consider your hypotheses for this project as listed below.

$H_0$: There is no difference in the average fare amount between customers who use credit cards and customers who use cash.

$H_A$: There is a difference in the average fare amount between customers who use credit cards and customers who use cash.



Your goal in this step is to conduct a two-sample t-test. Recall the steps for conducting a hypothesis test: 


1.   State the null hypothesis and the alternative hypothesis
2.   Choose a signficance level
3.   Find the p-value
4.   Reject or fail to reject the null hypothesis 



**Note:** For the purpose of this exercise, your hypothesis test is the main component of your A/B test. 

You choose 5% as the significance level and proceed with a two-sample t-test.

```python
# Filtrar datos solo de tipo de pago 1 (tarjeta) y 2 (efectivo)
credit_card_fares = taxi_data[taxi_data["payment_type"] == 1]["fare_amount"]
cash_fares = taxi_data[taxi_data["payment_type"] == 2]["fare_amount"]

# Realizar la prueba t de dos muestras (two-sample t-test)
t_statistic, p_value = stats.ttest_ind(credit_card_fares, cash_fares, equal_var=False)

# Mostrar resultados
print("🔍 Two-Sample T-Test Results:")
print("T-statistic:", t_statistic)
print("P-value:", p_value)

# Interpretación básica
alpha = 0.05
if p_value < alpha:
    print("✅ Result: Reject the null hypothesis — there is a significant difference in fare amounts.")
else:
    print("❌ Result: Fail to reject the null hypothesis — no significant difference detected.")
```

### ✅ Decision:

Based on the results of the two-sample t-test, the **p-value was [insert actual value here]**.

Since the p-value is **less than** the significance level of 0.05, we **reject the null hypothesis**.

This suggests that there **is a statistically significant difference** in the average fare amount between customers who pay with credit card and those who pay with cash.

### 📌 What assumptions had to be made, and why might this A/B test not be realistic?

To make causal claims in this A/B test, several **critical assumptions** were made:

1. **Random assignment**: We assumed that customers were randomly assigned to pay with credit card or cash, but in reality, **customers choose their payment method** based on preference or other factors (like income, trip type, or urgency).
   
2. **Independence of observations**: The test assumes that each ride is independent, but it's possible that frequent riders or driver-specific factors affect the data.
   
3. **No confounding variables**: The analysis does not control for trip distance, time of day, location, or tip amount, which might also explain differences in fare amount.

Because this is an educational dataset and not a controlled experiment, **we cannot make definitive causal conclusions** — only associations. A more rigorous randomized experiment would be needed for stronger causal claims.
