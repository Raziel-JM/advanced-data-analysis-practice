### Convert the `Date Joined` column to datetime

Use the `to_datetime()` function from the `pandas` library  to convert the `Date Joined` column to datetime. This splits each value into year, month, and date components. This is an important step in data cleaning, as it makes the data in this column easier to use in tasks you may encounter. To name a few examples, you may need to compare "date joined" between companies or determine how long it took a company to become a unicorn. ### Visualize the time it took companies to reach unicorn status

Visualize the longest time it took companies to reach unicorn status for each industry represented in the sample. To create a bar plot to visualize this, use the `bar()` function from the `matplotlib.pyplot` module. You'll first need to prepare the data.

```python
# Crear columna 'Years to Unicorn'
companies_sampled['Years to Unicorn'] = companies_sampled['Year Joined'] - companies_sampled['Year Founded']
```

```python
# Agrupar por industria y obtener el máximo de 'Years to Unicorn'
industry_time_to_unicorn = companies_sampled.groupby('Industry')['Years to Unicorn'].max()
```

```python
# Crear gráfico de barras
plt.bar(industry_time_to_unicorn.index, industry_time_to_unicorn.values)

# Establecer título
plt.title('Longest Time to Reach Unicorn Status by Industry')

# Establecer etiquetas de los ejes
plt.xlabel('Industry')
plt.ylabel('Time to Unicorn (Years)')

# Rotar las etiquetas en el eje x para evitar superposición
plt.xticks(rotation=45, ha='right')

# Mostrar el gráfico
plt.show()
```

### Visualize the maximum unicorn company valuation per industry

Visualize unicorn companies' maximum valuation for each industry represented in the sample. To create a bar plot to visualize this, use the `bar()` function from the `matplotlib.pyplot` module. Before plotting, create a new column that represents the companies' valuations as numbers (instead of strings, as they're currently represented). Then, use this new column to plot your data.

```python
# Function to convert valuation strings to numeric values
def convert_valuation(valuation):
    if isinstance(valuation, str):
        valuation = valuation.upper().replace(',', '').replace('$', '')  # Remove $, commas, and convert to uppercase
        if 'B' in valuation:
            return float(valuation.replace('B', '')) * 1e9  # Billion
        elif 'M' in valuation:
            return float(valuation.replace('M', '')) * 1e6  # Million
        elif 'K' in valuation:
            return float(valuation.replace('K', '')) * 1e3  # Thousand
        else:
            return float(valuation)  # For cases like '100'
    return valuation  # Return if it's already numeric
```

```python
# Apply the function to the 'Valuation' column and create a new numeric column
companies_sampled['Valuation_numeric'] = companies_sampled['Valuation'].apply(convert_valuation)
```

```python
# Count the number of unicorn companies per industry
industry_counts = companies_sampled.groupby('Industry').size()

```

```python
# Create a bar plot for the number of companies in each industry
import matplotlib.pyplot as plt

# Plot the bar chart
plt.bar(industry_counts.index, industry_counts.values)

# Set title
plt.title('Number of Unicorn Companies per Industry')

# Set x-axis label
plt.xlabel('Industry')

# Set y-axis label
plt.ylabel('Number of Unicorn Companies')

# Rotate x-axis labels to avoid overlap
plt.xticks(rotation=45, ha='right')

# Display the plot
plt.show()
```
