### Observe trends over time

Implement the structuring approach that you have identified to observe trends over time in the companies that became unicorns for the year that interests you.

````python
# Step 1: Filter dataset for companies that joined unicorn status in a specific year (e.g., 2015)
year_of_interest = 2015
companies_2015 = companies[companies['Year Joined'] == year_of_interest]

# Step 2: Create a new column for time interval (e.g., weeks) from the Date Joined column
companies_2015['Week Joined'] = companies_2015['Date Joined'].dt.strftime('%Y-W%V')

# Step 3: Group by the time interval and count the number of companies in each interval
companies_per_week = companies_2015.groupby('Week Joined').size().reset_index(name='Number of Companies')

# Step 4: Display the first few rows to confirm that the result is correct
companies_per_week.head()
````

**Question:** How would you structure the data to observe trends in the average valuation of companies from 2020 to 2021?  

To observe trends in the average valuation of companies from 2020 to 2021, you could follow a similar approach but focus on the Valuation column:

    Filter: First, filter the dataset for companies founded in 2020 and 2021.

    Create Time Intervals: You could group the data by year, month, or even quarter, depending on the level of granularity you want.

    Convert Valuation to Numeric: Since the Valuation column contains strings (e.g., "$100B"), you would need to clean the data, convert it to numeric, and strip away the dollar signs and "B".

    Group by Time Interval: After converting the Valuation column, group the data by the time interval (e.g., month or quarter), and calculate the average valuation for each group.

### Compare trends over time

Implement the structuring approach that you have identified in order to compare trends over time in the average valuation of companies that became unicorns in the year you selected above and in another year of your choice. Keep in mind the data type of the `Valuation` column and what the values in that column contain currently.

````python
# Step 1: Filter the dataset for two specific years (e.g., 2015 and 2018)
companies_2015 = companies[companies['Year Founded'] == 2015]
companies_2018 = companies[companies['Year Founded'] == 2018]

# Step 2: Concatenate the two subsets
companies_comparison = pd.concat([companies_2015, companies_2018], ignore_index=True)

# Step 3: Add a time interval column (e.g., Quarter)
companies_comparison['Quarter Joined'] = companies_comparison['Date Joined'].dt.to_period('Q').dt.strftime('%Y-Q%q')

# Step 4: Clean and transform the 'Valuation' column
companies_comparison['Valuation'] = companies_comparison['Valuation'].replace({'\$': '', 'B': ''}, regex=True).astype(float)

# Step 5: Group by the 'Quarter Joined' column and compute the average valuation
avg_valuation_by_quarter = companies_comparison.groupby(['Quarter Joined', 'Year Founded'])['Valuation'].mean().reset_index()

# Display the first few rows of the result
avg_valuation_by_quarter.head()
````

### Visualize the average valuation over the quarters

Using the subset of companies that joined in the years of interest, create a grouped bar plot to visualize the average valuation over the quarters, with two bars for each time interval. There will be two bars for each time interval. This allows you to compare quarterly values between the two years.

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Assuming 'companies' is the DataFrame containing the relevant data

# Step 1: Extract the year component and time interval (quarter) from the 'Date Joined' column
# We will assume 'Date Joined' is a datetime column and extract year and quarter
companies['Year Joined'] = companies['Date Joined'].dt.year
companies['Quarter Joined'] = companies['Date Joined'].dt.to_period('Q')

# Step 2: Filter companies for the years of interest (e.g., 2020 and 2021)
companies_filtered = companies[companies['Year Joined'].isin([2020, 2021])]

# Step 3: Convert 'Valuation' to a numeric column (removing $ and B, converting to float)
companies_filtered['Valuation'] = companies_filtered['Valuation'].replace({'\$': '', 'B': '', ',': ''}, regex=True).astype(float)

# Step 4: Group the data by 'Year Joined' and 'Quarter Joined' and calculate the average valuation
avg_valuation_by_quarter = companies_filtered.groupby(['Year Joined', 'Quarter Joined'])['Valuation'].mean().reset_index()

# Step 5: Set the size of the plot
plt.figure(figsize=(10, 6))

# Step 6: Create the grouped bar plot
sns.barplot(data=avg_valuation_by_quarter, x='Quarter Joined', y='Valuation', hue='Year Joined')

# Step 7: Set the title of the plot
plt.title('Average Valuation Over Quarters for 2020 and 2021')

# Step 8: Set the x-axis label
plt.xlabel('Quarter')

# Step 9: Set the y-axis label
plt.ylabel('Average Valuation (in Billion $)')

# Step 10: Display the plot
plt.show()
````
